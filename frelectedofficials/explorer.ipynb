```python
import httpx2
import io
import unidecode
import pandas
```

```python
url = 'https://static.data.gouv.fr/resources/repertoire-national-des-elus-1/20260811-154541/elus-conseiller-darrondissement-ca.csv'
```

```python
async with httpx2.AsyncClient() as client:
    response = await client.get(url)
    buffer = io.BytesIO(response.content)
    df = pandas.read_csv(buffer, sep=';')
```

```python
columns = df.columns.tolist()
```

```python
fixed_columns = [col.replace(' ', '_').replace("'", '').replace('-', '_').lower() for col in columns]
```

```python
fixed_columns = [unidecode.unidecode(col) for col in fixed_columns]
```
