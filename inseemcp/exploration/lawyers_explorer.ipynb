```python
import pandas
import pathlib
```

```python
# https://www.data.gouv.fr/datasets/annuaire-des-avocats-de-france?resource_id=99c8ff76-c0ff-4046-80d7-c10fc8e51c34
path = pathlib.Path('../data/lawyers.parquet')
df = pandas.read_parquet(path)
```

```python
df.head(2)
```

```python
df.Barreau.count()
```

```python
df.rename(columns={
    'civilit': 'civilite',
    'acSalarie': 'actif_salarie',
    'Barreau': 'barreau',
    'BarreauId': 'barreau_id',
    'avCnbfCode': 'av_cnbf_code',
    'avLANG': 'av_lang',
    'avNom': 'nom',
    'avPrenom': 'prenom',
    'avBarEntree': 'av_bar_entree',
    'cbAdresse1': 'cb_adresse_1',
    'cbAdresse2': 'cb_adresse_2',
    'cbCp': 'cb_cp',
    'cbVille': 'cb_ville',
    'cbTel': 'cb_tel',
    'cbFax': 'cb_fax',
    'cbSiretSiren': 'cb_siret_siren',
    'cbSiretNic': 'cb_siret_nic',
    'cbRaisonSociale': 'cb_raison_sociale',
    'acDateEntree': 'ac_date_entree',
    'cbFormJuri': 'cb_form_juri',
    'avMelOrdre': 'av_mel_ordre',
    'spLibelle1': 'sp_libelle_1',
    'spLibelle2': 'sp_libelle_2',
    'spLibelle3': 'sp_libelle_3',
    'acDateSerment': 'ac_date_serment',
    'avDateExer': 'av_date_exer',
    'avInscription': 'av_inscription',
    ' ': 'autre',
    ' .1': 'autre_1'
}, inplace=True)
```

```python
df = df[['civilite', 'actif_salarie', 'barreau', 'barreau_id', 'av_cnbf_code', 'av_lang', 'nom', 'prenom', 'av_bar_entree', 'cb_adresse_1', 'cb_adresse_2', 'cb_cp', 'cb_ville', 'cb_tel', 'cb_fax', 'cb_siret_siren', 'cb_siret_nic', 'cb_raison_sociale', 'ac_date_entree', 'cb_form_juri', 'av_mel_ordre', 'sp_libelle_1', 'sp_libelle_2', 'sp_libelle_3', 'ac_date_serment', 'av_date_exer', 'av_inscription']]
```

```python
df.to_parquet(pathlib.Path('../data/lawyers-cleaned.parquet'), index=False)
```
